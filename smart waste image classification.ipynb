```python

# ============================================================
# SMART WASTE IMAGE CLASSIFICATION
# DEEP LEARNING + CNN + MOBILENETV2 + GRADIO
# ============================================================

# 1. INSTALL LIBRARIES
!pip install -q tensorflow gradio scikit-learn matplotlib pillow

# 2. IMPORT LIBRARIES

import os
from pathlib import Path
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import gradio as gr

from PIL import Image as PILImage

from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.callbacks import EarlyStopping

print("TensorFlow Version:", tf.__version__)

# ============================================================
# 3. DATASET PATH - FIXED FOR JUPYTER
# ============================================================

from pathlib import Path

DATASET_DIR = Path("Garbage_Dataset")

if not DATASET_DIR.exists():
    raise FileNotFoundError(
        "Dataset folder not found. "
        "Please check ZIP_PATH and extraction."
    )

image_extensions = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

def has_images(folder):

    return any(
        f.suffix.lower() in image_extensions
        for f in folder.iterdir()
        if f.is_file()
    )

candidates = []

for folder in [DATASET_DIR] + list(DATASET_DIR.rglob("*")):

    if not folder.is_dir():
        continue

    class_folders = [
        sub for sub in folder.iterdir()
        if sub.is_dir() and has_images(sub)
    ]

    if len(class_folders) >= 2:
        candidates.append((folder, len(class_folders)))

if not candidates:
    raise ValueError(
        "Waste class folders not found. "
        "Check extracted dataset structure."
    )

dataset_path = max(
    candidates,
    key=lambda x: x[1]
)[0]

print("Dataset Folder:", dataset_path)
print("Number of Classes:", len([
    sub for sub in dataset_path.iterdir()
    if sub.is_dir() and has_images(sub)
]))
    

# ============================================================
# 4. LOAD IMAGE DATASET
# ============================================================

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

train_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int"
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int"
)

class_names = train_ds.class_names
NUM_CLASSES = len(class_names)

print("\nWaste Categories:", class_names)
print("Number of Classes:", NUM_CLASSES)

# ============================================================
# 5. DISPLAY SAMPLE IMAGES
# ============================================================

plt.figure(figsize=(12, 8))

for images, labels in train_ds.take(1):
    for i in range(min(9, len(images))):
        plt.subplot(3, 3, i + 1)

        plt.imshow(images[i].numpy().astype("uint8"))

        plt.title(class_names[int(labels[i])])
        plt.axis("off")

plt.tight_layout()
plt.show()

# ============================================================
# 6. DATA AUGMENTATION
# ============================================================

data_augmentation = models.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1)
])

# ============================================================
# 7. LOAD MOBILENETV2 CNN - OFFLINE FIX
# ============================================================

base_model = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights=None
)

# Freeze pretrained layers
base_model.trainable = True

print("MobileNetV2 CNN Loaded Successfully!")

# ============================================================
# 8. BUILD DEEP LEARNING CNN MODEL
# ============================================================

inputs = layers.Input(shape=(224, 224, 3))

x = data_augmentation(inputs)

# Normalize images for MobileNetV2
x = layers.Rescaling(
    scale=1./127.5,
    offset=-1
)(x)

# MobileNetV2 feature extraction
x = base_model(x, training=False)

# Additional CNN layer
x = layers.Conv2D(
    128,
    kernel_size=3,
    padding="same",
    activation="relu"
)(x)

x = layers.MaxPooling2D(pool_size=2)(x)

# Deep Learning layers
x = layers.GlobalAveragePooling2D()(x)

x = layers.Dense(
    256,
    activation="relu"
)(x)

x = layers.Dropout(0.4)(x)

x = layers.Dense(
    128,
    activation="relu"
)(x)

x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)

model = models.Model(inputs, outputs)

# ============================================================
# 9. COMPILE MODEL
# ============================================================

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# ============================================================
# 10. OPTIMIZE DATASET
# ============================================================

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(
    buffer_size=AUTOTUNE
)

val_ds = val_ds.prefetch(
    buffer_size=AUTOTUNE
)

# ============================================================
# 11. TRAIN MODEL
# ============================================================

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=4,
    restore_best_weights=True
)

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15,
    callbacks=[early_stop]
)

# ============================================================
# 12. MODEL EVALUATION
# ============================================================

loss, accuracy = model.evaluate(val_ds)

print("\n================================")
print("MODEL PERFORMANCE")
print("================================")

print("Validation Loss:", round(loss, 4))

print(
    "Validation Accuracy:",
    round(accuracy * 100, 2),
    "%"
)

# ============================================================
# 13. ACCURACY AND LOSS GRAPH
# ============================================================

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("CNN Model Accuracy")
plt.xlabel("Epochs")
plt.legend()

plt.subplot(1, 2, 2)

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("CNN Model Loss")
plt.xlabel("Epochs")
plt.legend()

plt.tight_layout()
plt.show()

# ============================================================
# 14. SAVE TRAINED MODEL
# ============================================================

model.save("smart_waste_cnn_model.keras")

print("\nModel Saved Successfully!")

# ============================================================
# 15. IMAGE PREDICTION FUNCTION
# ============================================================

def predict_waste(image):

    if image is None:
        return "Please upload a waste image."

    image = image.convert("RGB")
    image = image.resize(IMG_SIZE)

    img_array = np.array(
        image,
        dtype=np.float32
    )

    img_array = np.expand_dims(
        img_array,
        axis=0
    )

    prediction = model.predict(
        img_array,
        verbose=0
    )[0]

    predicted_index = np.argmax(prediction)

    predicted_class = class_names[predicted_index]

    confidence = prediction[predicted_index] * 100

    results = {
        class_names[i]: float(prediction[i])
        for i in range(NUM_CLASSES)
    }

    return (
        f"♻️ Predicted Waste: {predicted_class}\n"
        f"🎯 Confidence: {confidence:.2f}%"
    ), results

# ============================================================
# 16. GRADIO APPLICATION
# ============================================================

demo = gr.Interface(

    fn=predict_waste,

    inputs=gr.Image(
        type="pil",
        label="Upload Waste Image"
    ),

    outputs=[
        gr.Textbox(
            label="Waste Classification Result"
        ),
        gr.Label(
            label="Prediction Probabilities",
            num_top_classes=NUM_CLASSES
        )
    ],

    title="♻️ SMART WASTE CLASSIFICATION",

    description=(
        "AI Based Waste Classification Using "
        "Deep Learning, CNN and MobileNetV2. "
        "Upload a waste image to predict its category."
    )

)

# ============================================================
# 17. LAUNCH GRADIO
# ============================================================

demo.launch(share=True)
```
